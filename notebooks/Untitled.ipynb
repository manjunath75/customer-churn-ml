```python
# Next concept: Data Leakage

# This is one of the most important concepts for an ML Engineer.

# Suppose we have:

# Raw data
#    ↓
# Train/Test Split
#    ↓
# Preprocessing
#    ↓
# Model

# The test set must remain unseen while we're learning anything.

# For example, suppose we standardize a numerical feature.

# We calculate:

# mean
# standard deviation

# If we calculate those using all 7,043 customers, the test data has influenced the preprocessing.

# That is a form of data leakage.

# Instead:

# Training data
#      ↓
# calculate mean/std
#      ↓
# fit preprocessing
#      ↓
# transform training data

# Test data
#      ↓
# use the SAME mean/std
#      ↓
# transform test data

# The test set is transformed, but it does not teach the preprocessing step anything.

# The rule to remember

# Fit on training data. Transform both training and test data.

# Not:

# Fit on all data and then split.
```

```python
# Why this matters in your project

# We have some preprocessing decisions coming up:

# TotalCharges needs conversion to numeric.
# Missing values need handling.
# Categorical columns need encoding.
# Numerical features may need scaling.

# Some of these operations can learn information from the data.

# That's why we'll eventually use:

# ColumnTransformer
#        ↓
# Pipeline
#        ↓
# Model

# This allows us to package preprocessing + model together and greatly reduces the risk of accidentally leaking information.
```

```python
# Our preprocessing plan

# We're heading toward something like:

#                   Features
#                      │
#           ┌──────────┴──────────┐
#           ↓                     ↓
#      Numerical              Categorical
#           │                     │
#    ┌──────┴──────┐        One-hot encoding
#    ↓             ↓
# Missing       Scaling
# values
#    │
#    └──────────────┐
#                   ↓
#              Model
```

```python
# Suppose we encode:

# Month-to-month = 0
# One year       = 1
# Two year       = 2

# The problem is that we've accidentally told the model that these categories have a numerical distance and ordering:

# Two year (2) > One year (1) > Month-to-month (0)

# and that the difference between them is meaningful:

# 2 - 1 = 1
# 1 - 0 = 1

# For many categorical variables, that relationship doesn't exist.

# Example

# Consider:

# PaymentMethod

# Electronic check = 0
# Mailed check      = 1
# Bank transfer     = 2
# Credit card       = 3

# Does Credit card = 3 mean it's "three times" Electronic check?

# No.

# Does moving from 1 → 2 represent the same kind of change as 2 → 3?

# No.

# But a model receiving those numbers can interpret them as numerical values.

# One-hot encoding solves this

# Instead of:

# Contract = 0, 1, 2

# we create separate binary columns:

# Contract_Month-to-month
# Contract_One year
# Contract_Two year

# For example:

# Contract	Month-to-month	One year	Two year
# Month-to-month	1	0	0
# One year	0	1	0
# Two year	0	0	1

# Now the model doesn't assume that:

# Two year > One year > Month-to-month

# It simply sees three separate categories.

# The key concept to remember

# One-hot encoding is appropriate when categories are nominal and don't have a meaningful numerical order.

# There is another type called ordinal encoding, which can use numbers when the categories genuinely have an order, such as:

# Low → 0
# Medium → 1
# High → 2

# because here the ordering actually means something.
```

```python
# 1. What is scaling?

# Your numerical features have very different ranges:

# tenure          0 → 72
# MonthlyCharges  18.25 → 118.75
# TotalCharges    0 → several thousand

# If we feed these directly into some models, TotalCharges has much larger numerical values than the other features.

# Scaling transforms numerical features to a comparable scale.

# A common method is StandardScaler:

# z = x -  mue // sigma

# where:

# x = original value
# μ = mean calculated from training data
# σ = standard deviation calculated from training data

# After scaling, a feature roughly has:

# mean ≈ 0
# standard deviation ≈ 1
```

```python
# 2. Why does Logistic Regression care?

# Logistic Regression learns coefficients for features:

# Churn probability
#         ↑
#    coefficients
#         ↑
# tenure + MonthlyCharges + TotalCharges + ...

# If features have very different scales, optimization can behave poorly and coefficient magnitudes become harder to compare.

# Scaling generally makes the optimization process more well-behaved.

# So for our Logistic Regression, we'll use:

# StandardScaler()
```

```python
# 3. What about Decision Trees and Random Forest?

# This is an important distinction.

# Tree models make decisions like:

# MonthlyCharges <= 70.5 ?

# They care about the ordering and thresholds, not the numerical magnitude in the same way.

# Therefore:

# Decision Trees and Random Forests generally don't require feature scaling.

# So we have:

# Model	Scaling?
# Logistic Regression	Usually yes
# Decision Tree	Not required
# Random Forest	Not required
# Boosting trees	Usually not required

# This is one reason preprocessing should be designed with the model in mind.
```

```python
# 4. But our project needs multiple models

# We're going to compare:

# Logistic Regression
# Decision Tree
# Random Forest
# Boosting model

# Rather than manually creating different preprocessing code for every model, we'll use Pipelines.

# Conceptually:

# Logistic Regression Pipeline

# Raw data
#    ↓
# Convert TotalCharges
#    ↓
# Numerical preprocessing
#    ↓
# StandardScaler
#    ↓
# OneHotEncoder
#    ↓
# Logistic Regression

# For a tree model:

# Raw data
#    ↓
# Convert TotalCharges
#    ↓
# Numerical preprocessing
#    ↓
# OneHotEncoder
#    ↓
# Decision Tree

# The pipeline ensures that preprocessing is fitted correctly only using the training data.
```

```python
# One more important issue before coding

# Remember our 11 blank TotalCharges values?

# We need to decide what to do with them.

# We know:

# tenure = 0
# TotalCharges = blank

# This is where data cleaning meets domain knowledge.

# We have at least two reasonable possibilities:

# Option A: Treat blank TotalCharges as 0.

# Because a customer with zero months of tenure may genuinely have accumulated zero charges.

# Option B: Impute the missing value using a statistical method such as the median.

# But blindly replacing it with the median could create a value that doesn't make much sense for a customer whose tenure is zero.

# Think about this before we implement it:

# If:

# tenure = 0
# TotalCharges = blank

# would you choose A (0) or B (median), and why?

# There isn't just a coding answer here—the important part is your reasoning.
```

```python
# The strongest reasoning is not necessarily that they "might not have been billed during the 1st month." Rather:

# tenure = 0 indicates a newly started customer, and the blank TotalCharges values appear consistently for those customers. Since no tenure has accumulated, treating TotalCharges as 0 is consistent with the dataset's apparent domain meaning.

# This is a domain-based missing-value decision, rather than blindly applying a statistical imputation method.

# What we've learned here

# This is an important ML Engineering principle:

# Don't automatically use median/mean imputation just because a value is missing. First understand what the missing value means.

# For example:

# Missing age → median might be reasonable.
# Missing income → median might be reasonable depending on context.
# TotalCharges blank + tenure = 0 → 0 may make more domain sense.

# We should document this decision in the project so another engineer understands why we made it.
```

```python
# 2. Think of it as two processing branches
#                        X
#                        │
#                 ColumnTransformer
#                  /              \
#                 /                \
#        Numerical columns     Categorical columns
#               │                     │
#       StandardScaler          OneHotEncoder
#               │                     │
#               └─────────┬───────────┘
#                         ↓
#                  processed features

# This is much safer than manually preprocessing each column.
```

```python
numeric_features = ["tenure","MonthlyCharges","TotalCharges"]

categorical_features = ["SeniorCitizen","gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

preprocessor = ColumnTransformer(transformers=[("num",StandardScaler(),numeric_features),
                                               ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_features)])
```

```python
# Two things to understand
# StandardScaler()
# ("num", StandardScaler(), numerical_features)

# means:

# Take these numerical columns and standardize them.

# OneHotEncoder(handle_unknown="ignore")
# ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)

# means:

# Convert these categorical columns into one-hot encoded columns.

# handle_unknown="ignore" is particularly useful in production.

# Imagine the model was trained and saw:

# PaymentMethod:
# - Electronic check
# - Mailed check
# - Bank transfer
# - Credit card

# Later, an API request contains a previously unseen category.

# Instead of crashing because the category wasn't present during training, the encoder can handle it without producing an error.
```

```python
from sklearn.pipeline import Pipeline 
from sklearn.linear_model import LogisticRegression


logistic_pipeline = Pipeline(steps=[("preprocessor",preprocessor),
                                    ("model",LogisticRegression(max_iter=1000))
                                   ])
```

```python
# This result is useful because it shows that the unrestricted Decision Tree is performing worse than our Logistic Regression on the test set.

# 1. Compare the two models
# Metric	Logistic Regression	Decision Tree
# Accuracy	80.55%	71.75%
# Churn precision	65.7%	47%
# Churn recall	55.9%	50%
# Churn F1	60%	48%
# ROC-AUC	0.842	0.648

# The important point is that we're not choosing a model based only on accuracy.

# For this churn problem, the Yes class is the important class, so we also care about:

# How many actual churners we detect → recall
# How many predicted churners really churn → precision
# Overall balance → F1
# Ability to rank churners above non-churners across thresholds → ROC-AUC

# The tree is lower on all of these in this test.

# 2. Why did the tree perform poorly?

# Remember how we created it:

# DecisionTreeClassifier(random_state=42)

# We didn't tell it how deep it should be.

# A Decision Tree can keep splitting:

# Root
#  ├── split
#  │    ├── split
#  │    │    ├── split
#  │    │    └── split
#  │    └── split
#  └── split
#       ├── split
#       └── split

# Eventually, it can create very small leaves that describe the training data extremely specifically.

# This is the classic overfitting problem.

# The fact that your tree produced probabilities such as:

# [1.0, 0.0]
# [0.0, 1.0]

# was an early warning sign.

# But notice: we haven't actually proven overfitting yet.

# To demonstrate it properly, we should compare training performance vs test performance.
```

```python
tree_train_pred = tree_pipeline.predict(X_train)
tree_train_prob = tree_pipeline.predict_proba(X_train)

print("Training Accuracy:",accuracy_score(y_train,tree_train_pred))

print("Training ROC-AUC:",roc_auc_score((y_train=="Yes").astype(int),tree_train_prob[:,1]))
```

```python
print(classification_report(y_train, tree_train_pred))
```

```python
# The important ML concept: model complexity

# Think of max_depth as controlling how complicated the tree is.

# Low max_depth
#       ↓
# Simpler tree
#       ↓
# May underfit

# High max_depth
#       ↓
# More complicated tree
#       ↓
# May overfit
```

```python
# Our current tree effectively has no meaningful restriction, which is why it reached ~99.8% training accuracy.

# We now want to find a middle ground:

# Too simple       Appropriate       Too complex
#     │                 │                  │
# Underfitting      Generalizes        Overfitting

# This is a fundamental bias–variance tradeoff.

# High bias → model is too simple → underfitting.
# High variance → model reacts too strongly to training data → overfitting.

# Your current tree is showing the high-variance/overfitting side.
```

```python
tree_pipeline_depth3 = Pipeline(steps=[("preeprocessor",tree_preprocessor),("model",DecisionTreeClassifier(max_depth=3,random_state=42))])
```

```python
tree_pipeline_depth3.fit(X_train,y_train)
```

```python
train_pred = tree_pipeline_depth3.predict(X_train)
test_pred = tree_pipeline_depth3.predict(X_test)

print("Training Accuracy ",accuracy_score(y_train,train_pred))
print("Test Accuracy",accuracy_score(y_test,test_pred))
```

```python
test_prob = tree_pipeline_depth3.predict_proba(X_test)

print("Test ROC-AUC:",
      roc_auc_score(
          (y_test == "Yes").astype(int),
          test_prob[:, 1]
      ))
```

```python
print(confusion_matrix(y_test, test_pred))

print(classification_report(y_test, test_pred))
```

```python
# 1. What is a boosting model?

# Boosting is an ensemble learning technique where we build many weak models, usually small decision trees, sequentially.

# The key idea is:

# Each new tree tries to improve the mistakes made by the previous trees.

# Think of it like this:

# Training data
#      ↓
# Tree 1
#      ↓
# Find what Tree 1 gets wrong
#      ↓
# Tree 2 focuses on those errors
#      ↓
# Tree 3 focuses on remaining errors
#      ↓
# ...
#      ↓
# Final prediction

# The individual trees are usually simple, but their combined prediction can be very powerful.

# 2. How is this different from Random Forest?

# This is an important interview question.

# Random Forest

# Random Forest builds trees independently/in parallel.

#              Data
#               ↓
#        ┌──────┼──────┐
#        ↓      ↓      ↓
#      Tree1  Tree2  Tree3
#        ↓      ↓      ↓
#        └──────┼──────┘
#               ↓
#        Majority vote

# Each tree doesn't specifically try to fix the previous tree.

# Boosting

# Boosting builds trees sequentially.

# Data
#  ↓
# Tree 1
#  ↓
# Errors
#  ↓
# Tree 2
#  ↓
# Remaining errors
#  ↓
# Tree 3
#  ↓
# ...
#  ↓
# Final prediction

# So:

# Random Forest	Boosting
# Trees are independent	Trees are sequential
# Focuses on averaging many trees	Focuses on correcting errors
# Reduces variance	Can reduce bias and capture complex patterns
# Usually robust	Often very powerful
# Less sensitive to tuning	Can be more sensitive to tuning
# 3. Why would we use boosting?

# Suppose your Logistic Regression is doing this:

# Customer characteristics
#        ↓
# Linear relationship
#        ↓
# Churn probability

# But the actual relationship might be more complicated.

# For example:

# IF tenure is low
# AND contract is month-to-month
# AND monthly charges are high
# → high churn probability

# That's an interaction between multiple features.

# A boosting model can learn these kinds of nonlinear relationships and interactions automatically.

# This is one reason boosting models are extremely common for tabular business data such as:

# Customer churn
# Fraud detection
# Credit risk
# Loan default
# Insurance prediction
# Customer conversion
# Sales prediction
# 4. What exactly happens during boosting?

# Imagine our first tree makes these mistakes:

# Actual churners:       100
# Tree 1 detects:         60
# Missed:                 40

# The next tree pays more attention to the difficult examples.

# Tree 1
#   ↓
# identify errors
#   ↓
# Tree 2
#   ↓
# correct some errors
#   ↓
# Tree 3
#   ↓
# correct remaining errors

# The final model combines the contributions of all those trees.

# This is the fundamental idea behind algorithms such as:

# Gradient Boosting
# XGBoost
# LightGBM
# CatBoost
# HistGradientBoosting
# 5. Why is it called gradient boosting?

# This is slightly more advanced, but important for an ML Engineer interview.

# The model isn't simply saying:

# "Tree 2, fix whatever Tree 1 got wrong."

# Instead, gradient boosting calculates how the current predictions should change to reduce the loss function.

# Conceptually:

# Current prediction
#        ↓
# Calculate loss/error
#        ↓
# Calculate direction of improvement
#        ↓
# Train another tree toward that improvement
#        ↓
# Update model
#        ↓
# Repeat

# The "gradient" refers to the direction in which the loss can be reduced.

# You don't need to derive the mathematics yet. The important conceptual distinction is:

# Gradient boosting sequentially adds trees that move the model toward lower loss.

# 6. When would YOU use boosting?

# For your churn project, we're using it because we want to answer:

# "Can a nonlinear boosting model outperform our Logistic Regression and Random Forest on this tabular dataset?"

# Your current comparison is:

# Logistic Regression
#         ↓
# Decision Tree
#         ↓
# Random Forest
#         ↓
# Boosting
#         ↓
# Compare using the same evaluation process

# Your Logistic Regression currently has:

# CV ROC-AUC ≈ 0.846

# So boosting isn't automatically "better."

# We have to test it.

# If boosting gives:

# ROC-AUC = 0.86

# while maintaining useful recall/precision, it may become our final candidate.

# If it gives:

# ROC-AUC = 0.81

# then we have evidence that Logistic Regression is actually better for this dataset.

# That's an important ML engineering lesson:

# More complex ≠ automatically better.

# 7. One more important distinction

# There are two meanings of "boosting" you should know.

# AdaBoost

# Older boosting approach:

# Tree 1 → increase weight of misclassified samples
# Tree 2 → focus more on those samples
# Tree 3 → ...
# Gradient Boosting

# More modern approach:

# Tree 1 → calculate residual/loss
# Tree 2 → learn to reduce that loss
# Tree 3 → reduce remaining loss
# ...

# And then we have highly optimized implementations such as XGBoost, which became extremely popular for structured/tabular ML.

# For your project, we can use XGBoost if available, or HistGradientBoostingClassifier from scikit-learn.

# For your learning path

# I'd recommend we learn the basic Gradient Boosting idea first, then implement the boosting model in your churn project.

# That way you understand why you're writing the code rather than just adding another algorithm to the notebook.
```

```python
# Lesson: Understanding Gradient Boosting

# Let's start with the core idea before writing any code.

# 1. The problem boosting tries to solve

# In your churn project, you already trained a Decision Tree. It made some correct predictions and some incorrect ones.

# Suppose we have this simplified training data:

# Customer

	

# Actual churn

	

# Tree 1 prediction




# A

	

# Yes

	

# No




# B

	

# No

	

# No




# C

	

# Yes

	

# Yes




# D

	

# Yes

	

# No




# E

	

# No

	

# Yes

# Tree 1 makes mistakes on customers A, D and E.

# A boosting model tries to improve the existing model by learning from its errors.

# Tree 1

# Makes initial predictions

# Calculate what needs improvement

# Identify the direction that reduces prediction loss

# Tree 2, then Tree 3, and so on

# Each new tree contributes to improving the combined model

# One important detail: gradient boosting does not simply memorize every misclassified customer. It uses the gradient of a loss function to determine how the model's predictions should improve.
```

```python
# 2. How does the model combine the trees?

# Unlike a Random Forest, which combines predictions from independently trained trees, gradient boosting builds an additive model.

# Conceptually:

# F new(x)=F old(x)+ηh(x)

# Where:

# F
# old
# 	​

# (x): current model prediction.

# h(x): contribution learned by the new tree.

# η: learning rate, which controls how much the new tree contributes.

# A smaller learning rate makes each tree's contribution smaller, often requiring more trees.
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

boosting_preprocessor = ColumnTransformer(transformers=[("num","passthrough",numeric_features),("cat",OneHotEncoder(handle_unknown="ignore",sparse_output=False),categorical_features)])
```

```python
boosting_pipeline  = Pipeline(steps=[("preprocessor",boosting_preprocessor),("model",HistGradientBoostingClassifier(random_state=42))])
```

```python
boosting_pipeline.fit(X_train,y_train)
```
